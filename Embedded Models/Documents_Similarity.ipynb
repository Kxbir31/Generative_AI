```python
from langchain_huggingface import HuggingFaceEmbeddings
from dotenv import load_dotenv
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np
```

```python
load_dotenv()
```

```python
embeddings = HuggingFaceEmbeddings(model = 'sentence-transformers/all-MiniLM-L6-v2')

```

```python
document =["Virat Kohli – One of the greatest modern batsmen, known for his consistency, aggressive mindset, and record-breaking performances.",
"Rohit Sharma – Famous for his elegant batting style and the highest individual score (264) in One Day Internationals.",
"Jasprit Bumrah – Widely regarded as one of the world's best fast bowlers due to his unique action and exceptional accuracy.",
"Ben Stokes – An outstanding all-rounder known for delivering match-winning performances in high-pressure situations.",
"Babar Azam – Renowned for his technically sound batting, elegant stroke play, and consistency across formats."]
```

```python
query = "Tell me about Jasprit"
```

```python
doc_embeddings = embeddings.embed_documents(document)
query_embedding = embeddings.embed_query(query)
```

```python
scores = cosine_similarity([query_embedding], doc_embeddings)[0]
print(scores)
```

```python
index , score  = (sorted(list(enumerate(scores)),key=lambda x: x[1])[-1])
```

```python
print(query)
print(document[index])
print("cosine similarity is : ",scores)
```
