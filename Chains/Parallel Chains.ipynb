```python
from tempfile import template
from langchain_huggingface import ChatHuggingFace , HuggingFaceEndpoint
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_core.runnables import RunnableParallel
```

```python
from dotenv import load_dotenv
load_dotenv()
```

```python
import os

llm = HuggingFaceEndpoint(
    repo_id="deepseek-ai/DeepSeek-V4-Pro",
    task="text-generation",
    huggingfacehub_api_token=HUGGINGFACEHUB_ACCESS_TOKEN,
    temperature=0.8
)
```

```python

llm2 = HuggingFaceEndpoint(
    repo_id="Qwen/Qwen3-4B-Instruct-2507",
    task="text-generation",
    huggingfacehub_api_token=HUGGINGFACEHUB_ACCESS_TOKEN,
    temperature=0.8
)
```

```python
model1 = ChatHuggingFace(llm = llm)
model2 = ChatHuggingFace(llm = llm2)
```

```python
prompt1 = PromptTemplate(
    template='Generate simple notes from the following \n {text}',
    input_variables=['text']
)
```

```python
prompt2 = PromptTemplate(
    template='Generate a quiz following \n {text}',
    input_variables=['text']
)
```

```python
prompt3 = PromptTemplate(
    template ='Merge the notes and quiz into a single output but seprated \n - {notes} and {quiz}',
    input_variables=['notes', 'quiz']
)
```

```python
parser = StrOutputParser()
```

```python
parallel_chains = RunnableParallel({
    'notes': prompt1 | model1 | parser,
    'quiz' : prompt2 | model2 | parser,

})
```

```python
merge_chain = prompt3 | model1 | parser
```

```python
chain = parallel_chains | merge_chain
```
